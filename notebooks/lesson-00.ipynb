## 5. Lists and indexing

````python
prices = [58, 82, 55, 88, 96]
print(prices[0], prices[1], prices[-1])   # first, second, last
print(len(prices))                          # how many items
````

## 7. Making decisions and repeating: if and for

````python
price = 96
if price > 90:
    print("expensive")
elif price > 60:
    print("mid-range")
else:
    print("affordable")
````

````python
prices = [58, 82, 55, 88]
total = 0
for p in prices:
    total = total + p
print("total:", total, " average:", total / len(prices))
````

````python
for step in range(3):
    print("step", step)
squares = [x ** 2 for x in [1, 2, 3, 4]]
print(squares)
````

## 8. Functions

````python
def mse(y_true, y_pred):
    total = 0
    for t, p in zip(y_true, y_pred):   # zip walks through both lists together
        total += (p - t) ** 2
    return total / len(y_true)

print(mse([3, 5, 7], [2.5, 5, 8]))
````

## 10. NumPy: numbers in bulk

````python
x = np.array([1.0, 2.0, 3.0, 4.0])
print(x * 2)          # every element doubled
print(x + 10)
print(x ** 2)
print(x.mean(), x.sum(), x.max())
````

````python
X = np.array([[120, 3, 4.0],
              [ 95, 2, 6.5],
              [150, 4, 12.0]])   # 3 houses × 3 features
print(X.shape)
print(X[0])           # the first row (house 1)
print(X[:, 0])        # the first column (sizes)
print(X.mean(axis=0)) # the average of each column
````

````python
sizes = X[:, 0]
print(sizes > 100)
print(sizes[sizes > 100])
w = np.array([0.5, 2.0, -1.0])
print(X @ w)          # one prediction per house
````

## 11. pandas: tables of data

````python
df = pd.read_csv("../data/houses.csv")
print(df.shape)                # (rows, columns)
print(df.columns.tolist())
````

````python
print(df[["area", "size_m2", "price_lakh"]].head(4))   # the first 4 rows of three columns
````

````python
print(df["price_lakh"].mean())
big = df[df["size_m2"] > 120]
print(len(big), "houses over 120 m²")
print(big[["id", "size_m2", "price_lakh"]])
````

````python
print(df.groupby("area")["price_lakh"].mean().round(1))
df["price_per_m2"] = df["price_lakh"] / df["size_m2"]
print(df.sort_values("price_per_m2", ascending=False)[["id", "area", "price_per_m2"]].head(3).round(3))
````
