## 9. Your first ML code

````python
from sklearn.linear_model import LinearRegression

# features: house size in 100 m² (each row is one house)
X = [[1], [2], [3], [4], [5]]
# labels: price in units of 100,000 USD
y = [2.1, 3.9, 6.2, 7.8, 10.1]

model = LinearRegression()   # 1. choose a model
model.fit(X, y)              # 2. train: find the best w and b

print(model.coef_, model.intercept_)   # the learned w and b
print(model.predict([[6]]))            # 3. predict a new 600 m² house
````
